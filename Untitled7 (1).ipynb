```python
import glob

files = sorted(glob.glob("Big-Data-Analytics/datasets/text-files/**/*.txt", recursive=True))
print("Files found:", len(files))
print(*files, sep="\n")
```

```python

import re
from collections import Counter
from concurrent.futures import ThreadPoolExecutor

# CHANGE THESE NUMBERS
num_mappers = 5
num_reducers = 2

# Select files for mapper tasks
selected_files = files[:num_mappers]

# Map
def mapper(file):
    with open(file, encoding="utf-8", errors="ignore") as f:
        words = re.findall(r"\b\w+\b", f.read().lower())
    return [(w, 1) for w in words]

# Parallel Mappers
with ThreadPoolExecutor(max_workers=num_mappers) as pool:
    mapped = list(pool.map(mapper, selected_files))

# Combine + Partition
parts = [[] for _ in range(num_reducers)]
for pairs in mapped:
    for w, n in Counter(w for w, _ in pairs).items():
        parts[sum(map(ord, w)) % num_reducers].append((w, n))

# Shuffle + Sort + Reduce
def reducer(i):
    counts = Counter()
    for w, n in sorted(parts[i]):
        counts[w] += n

    with open(f"output{i}.txt", "w") as f:
        for w, n in sorted(counts.items()):
            f.write(f"{w} {n}\n")

# Parallel Reducers
with ThreadPoolExecutor(max_workers=num_reducers) as pool:
    list(pool.map(reducer, range(num_reducers)))

print("Mapper tasks:", len(selected_files))
print("Reducers:", num_reducers)
```

```python
for i in range(2):
    print(f"\noutput{i}.txt:")
    with open(f"output{i}.txt") as f:
        for line in list(f)[:20]:
            print(line.strip())
```

```python
total = sum(len(pairs) for pairs in mapped)
reduced = sum(
    sum(int(line.split()[-1]) for line in open(f"output{i}.txt"))
    for i in range(2)
)

print("Words from mappers:", total)
print("Words from reducers:", reduced)
print("Correct:", total == reduced)
```

```python
# Combine both reducer outputs into one file
results = []

for i in range(2):
    with open(f"output{i}.txt") as f:
        for line in f:
            word, count = line.strip().rsplit(" ", 1)
            results.append((word, int(count)))

# Sort and save
with open("final_output.txt", "w") as f:
    for word, count in sorted(results):
        f.write(f"{word} {count}\n")

print("Combined output:")
print(open("final_output.txt").read()[:2000])
```
